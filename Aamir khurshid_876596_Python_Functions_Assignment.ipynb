**Problem:**  
A teacher has a list of marks of all students in a class and wants key statistics.

````python
# Write your code below

def marks_processor(marks):
    highest = marks[0]
    lowest = marks[0]
    total = 0

    for mark in marks:
        total = total + mark

        if mark > highest:
            highest = mark

        if mark < lowest:
            lowest = mark

    average = total / len(marks)

    return highest, lowest, total, average


highest, lowest, total, average = marks_processor([80, 65, 90, 75, 85])

print("Highest Marks:", highest)
print("Lowest Marks:", lowest)
print("Total Marks:", total)
print("Average Marks:", average)
````

**Problem:**  
Salary calculation has several smaller steps. Use nested functions to organize the logic cleanly.

**Requirements:**
- Create an outer function called `calculate_salary()`.

  - Calculating allowance
  - Calculating deduction
  - Calculating final salary
- The outer function should combine the results and return the final salary.
- Call the outer function and display the result.

**Focus:** Nested functions

````python
def calculate_salary(basic_salary):

    def calculate_allowance():
        return basic_salary * 0.20

    def calculate_deduction():
        return basic_salary * 0.10

    def final_salary():
        allowance = calculate_allowance()
        deduction = calculate_deduction()

        return basic_salary + allowance - deduction

    return final_salary()


# Calling the outer function
salary = calculate_salary(50000)

print("Final Salary:", salary)

````

## Question 24 — Nested Order Processing

**Problem:**  
An order processing system calculates subtotal, discount, tax, and final amount. Organize this using nested functions.

**Requirements:**
- Create an outer function called `process_order()`.
- Inside it, create nested functions for:
  - Calculate subtotal
  - Calculate discount
  - Calculate tax
  - Calculate final amount
- Return the final order amount from the outer function.
- Demonstrate with sample values.

**Focus:** Nested functions + arguments + `return` + real-world logic

````python
# Write your code below
def process_order(price, quantity):

    def calculate_subtotal():
        return price * quantity

    def calculate_discount():
        subtotal = calculate_subtotal()
        return subtotal * 0.10

    def calculate_tax():
        subtotal = calculate_subtotal()
        return subtotal * 0.05

    def calculate_final_amount():
        subtotal = calculate_subtotal()
        discount = calculate_discount()
        tax = calculate_tax()

        return subtotal - discount + tax

    return calculate_final_amount()


# Calling the outer function
final_amount = process_order(1000, 2)

print("Final Order Amount:", final_amount)
````

## Question 29 — Student Result Management System

**Problem:**  
Create a complete student result system using functions.

**The system should:**
1. Accept student information
2. Accept multiple subject marks
3. Calculate total
4. Calculate percentage
5. Determine grade
6. Determine pass/fail
7. Display a final result card

**Requirements:**
- Use **multiple functions**
- Use arguments and `return`
- Use lists
- Use conditions and loops
- Use at least one advanced concept (`*args`, `**kwargs`, nested functions, or lambda)
- Do **not** write the entire program inside one function

**Focus:** Combining multiple function concepts

````python
# Write your code below
# 1. Student Information
def student_info(name, roll_no):
    return name, roll_no


# 2. Accept Subject Marks
def get_marks():
    marks = []

    subjects = ["English", "Math", "Computer", "Physics", "Chemistry"]

    for subject in subjects:
        mark = int(input("Enter " + subject + " marks: "))
        marks.append(mark)

    return marks


# 3. Calculate Total Using *args
def calculate_total(*marks):
    return sum(marks)


# 4. Calculate Percentage
def calculate_percentage(total):
    return (total / 500) * 100


# 5. Determine Grade
def calculate_grade(percentage):
    if percentage >= 80:
        return "A"
    elif percentage >= 70:
        return "B"
    elif percentage >= 60:
        return "C"
    elif percentage >= 50:
        return "D"
    else:
        return "F"


# 6. Determine Pass or Fail
def check_result(marks):
    for mark in marks:
        if mark < 40:
            return "Fail"

    return "Pass"


# 7. Display Final Result Card
def result_card(name, roll_no, marks, total, percentage, grade, result):

    subjects = ["English", "Math", "Computer", "Physics", "Chemistry"]

    print("\n========== STUDENT RESULT CARD ==========")

    print("Name:", name)
    print("Roll Number:", roll_no)

    print("-----------------------------------------")

    for i in range(len(subjects)):
        print(subjects[i], ":", marks[i])

    print("-----------------------------------------")

    print("Total Marks:", total, "/ 500")
    print("Percentage:", percentage, "%")
    print("Grade:", grade)
    print("Result:", result)

    print("=========================================")


# Main Program

name, roll_no = student_info("Ali", 101)

marks = get_marks()

total = calculate_total(*marks)

percentage = calculate_percentage(total)

grade = calculate_grade(percentage)

result = check_result(marks)

result_card(name, roll_no, marks, total, percentage, grade, result)
````

````python

# Customer Information Function
def customer_info(name, phone, address="Not Provided"):
    customer = {
        "name": name,
        "phone": phone,
        "address": address
    }
    return customer


# Product Details
def show_products():
    products = {
        "Laptop": 80000,
        "Mouse": 1500,
        "Keyboard": 2500,
        "Headphones": 3000,
        "USB": 1000
    }

    print("\n------ AVAILABLE PRODUCTS ------")

    for name, price in products.items():
        print(name, ":", price)

    return products


# Calculate Product Price
def product_total(price, quantity):
    return price * quantity


# Calculate Subtotal Using *args
def calculate_subtotal(*totals):
    subtotal = 0

    for total in totals:
        subtotal += total

    return subtotal


# Calculate Discount Using Nested Function
def calculate_discount(subtotal):

    def discount_rate():
        if subtotal >= 50000:
            return 10
        elif subtotal >= 20000:
            return 5
        else:
            return 0

    discount = subtotal * discount_rate() / 100
    return discount


# Calculate Tax
def calculate_tax(amount, tax_rate=5):
    return amount * tax_rate / 100


# Calculate Delivery Charges
def delivery_charges(subtotal):
    if subtotal >= 10000:
        return 0
    else:
        return 300


# Display Order Summary
def order_summary(customer, order, subtotal, discount, tax, delivery):

    final_bill = subtotal - discount + tax + delivery

    print("\n========== ORDER SUMMARY ==========")

    print("Customer Name:", customer["name"])
    print("Phone Number:", customer["phone"])
    print("Address:", customer["address"])

    print("\n------ ORDER DETAILS ------")

    for item in order:
        print(item["name"], "x", item["quantity"],
              "=", item["total"])

    print("\nSubtotal:", subtotal)
    print("Discount:", discount)
    print("Tax:", tax)
    print("Delivery Charges:", delivery)

    print("-----------------------------------")
    print("FINAL BILL:", final_bill)
    print("===================================")


# Main Program

name = input("Enter Customer Name: ")
phone = input("Enter Phone Number: ")
address = input("Enter Address: ")

customer = customer_info(name, phone, address)

products = show_products()

order = []
totals = []

while True:

    product_name = input("\nEnter Product Name (or done to finish): ")

    if product_name.lower() == "done":
        break

    if product_name in products:

        quantity = int(input("Enter Quantity: "))

        if quantity > 0:

            price = products[product_name]

            total = product_total(price, quantity)

            order.append({
                "name": product_name,
                "quantity": quantity,
                "total": total
            })

            totals.append(total)

            print("Product Added Successfully!")

        else:
            print("Quantity must be greater than zero.")

    else:
        print("Product Not Available!")


if len(order) > 0:

    subtotal = calculate_subtotal(*totals)

    discount = calculate_discount(subtotal)

    amount_after_discount = subtotal - discount

    tax = calculate_tax(amount_after_discount)

    delivery = delivery_charges(subtotal)

    # Lambda Function
    round_amount = lambda amount: round(amount, 2)

    tax = round_amount(tax)

    order_summary(
        customer,
        order,
        subtotal,
        discount,
        tax,
        delivery
    )

else:
    print("\nNo Products Ordered!")
# Write your code below
````
